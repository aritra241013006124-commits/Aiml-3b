````python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

print("BFS Traversal:")
bfs(graph, 'A')
````

````python

# Graph using Adjacency List
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# ---------------- BFS ----------------
def bfs(graph, start):
    visited = []
    queue = [start]

    while queue:
        node = queue.pop(0)

        if node not in visited:
            visited.append(node)
            queue.extend(graph[node])

    return visited


# ---------------- DFS ----------------
def dfs(graph, start, visited=None):
    if visited is None:
        visited = []

    visited.append(start)

    for node in graph[start]:
        if node not in visited:
            dfs(graph, node, visited)

    return visited


# Perform BFS and DFS
bfs_order = bfs(graph, 'A')
dfs_order = dfs(graph, 'A')

# Print traversal orders
print("BFS Traversal:", " -> ".join(bfs_order))
print("DFS Traversal:", " -> ".join(dfs_order))


# ---------------- Draw BFS Tree ----------------
import matplotlib.pyplot as plt
import networkx as nx

def draw_tree(order, title):
    G = nx.DiGraph()

    # Add nodes in traversal order
    for i, node in enumerate(order):
        G.add_node(node, order=i + 1)

    # Create edges according to the original tree
    for parent in graph:
        for child in graph[parent]:
            if child in order:
                G.add_edge(parent, child)

    # Positions for the tree
    pos = {
        'A': (0, 2),
        'B': (-1, 1),
        'C': (1, 1),
        'D': (-1.5, 0),
        'E': (-0.5, 0),
        'F': (1, 0)
    }

    labels = {node: f"{node}\n({order.index(node) + 1})"
              for node in order}

    nx.draw(
        G,
        pos,
        with_labels=True,
        labels=labels,
        node_size=2000,
        node_color="lightblue",
        font_size=12,
        arrows=False
    )

    plt.title(title)
    plt.show()


# Draw BFS tree
draw_tree(bfs_order, "BFS Traversal Tree")

# Draw DFS tree
draw_tree(dfs_order, "DFS Traversal Tree")
```

````
