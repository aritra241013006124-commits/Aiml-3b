```python
from collections import deque

def water_jug():
    capacity_a = 4
    capacity_b = 3
    target = 2

    start = (0, 0)

    queue = deque([(start, [])])
    visited = set()

    while queue:
        (a, b), path = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))
        path = path + [(a, b)]

        # Check target
        if a == target or b == target:
            print("Minimum sequence of steps:")
            for i, state in enumerate(path):
                print(i, ":", state)
            return

        states = []

        # Fill Jug A
        states.append((capacity_a, b))

        # Fill Jug B
        states.append((a, capacity_b))

        # Empty Jug A
        states.append((0, b))

        # Empty Jug B
        states.append((a, 0))

        # Pour A -> B
        amount = min(a, capacity_b - b)
        states.append((a - amount, b + amount))

        # Pour B -> A
        amount = min(b, capacity_a - a)
        states.append((a + amount, b - amount))

        for state in states:
            if state not in visited:
                queue.append((state, path))


water_jug()
```
